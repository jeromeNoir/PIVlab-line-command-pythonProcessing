```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib qt
```

```python
t=np.linspace(0, 120, 10000)
R=138e-3
k0=6
dphideg=2
flib=2.2

dphi=dphideg*np.pi/180 #rad  
wlib=2*np.pi*flib
k=k0*np.pi/R
l=2*R/k0

U0=dphi*R*wlib*np.cos(wlib*t) #m/s
wint=k*U0 #rad/s

p=np.cos(wint*t) + np.sin(wlib*t) #rad
```

```python
plt.figure()
plt.plot(t, p)
```

```python
dt = t[1] - t[0]
N = p.size

P = np.fft.rfft(p) / N
f = np.fft.rfftfreq(N, dt)

plt.figure()
plt.plot(f, np.abs(P))
plt.axvline(flib, color='r', ls='--', label=r'$f_{lib}$')
plt.axvline(flib / 2, color='g', ls='--', label=r'$f_{lib}/2$')
plt.xlim(0, 5 * flib)
plt.yscale('log')
plt.xlabel('frequency [Hz]')
plt.ylabel('|FFT(p)|')
plt.legend()
plt.show()
```
